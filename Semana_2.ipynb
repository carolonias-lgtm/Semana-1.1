```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Carrega o arquivo criando a variável 'df'
df = pd.read_excel('Dados_semana2.xlsx', header=1)

# 2. Faz a limpeza dos dados
df['Marcas'] = df['Marcas'].ffill()
df['Marcas'] = df['Marcas'].str.replace('\n', ' ')
df = df.dropna(subset=['Sabores'])

# 3. Exibe os dados prontos
df

```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Marcas', y='Valor energético (Kcal)', hue='Sabores', palette='tab10')
plt.title('Valor Energético (Kcal/100ml) por Marca e Sabor', fontsize=14, fontweight='bold')
plt.xlabel('Marca', fontsize=12)
plt.ylabel('Valor Energético (Kcal)', fontsize=12)
plt.xticks(rotation=20, ha='right')
plt.legend(title='Sabor', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_excel('Dados_semana2.xlsx', header=1)
df = df.dropna(subset=['Sabores'])
contagem = df['Sabores'].value_counts()
print(contagem)
```

```python

plt.figure(figsize=(8, 8))
plt.pie(contagem, labels=contagem.index, autopct='%1.0f%%')
plt.title('Porcentagem de Produtos por Sabor na Tabela')
plt.show()
```

```python

plt.figure(figsize=(9, 5))
plt.hist(df['Valor energético (Kcal)'], bins=6, color='skyblue', edgecolor='black')
plt.title('Distribuição de Calorias (Kcal/100ml) nas Bebidas Proteicas')
plt.xlabel('Valor Energético (Kcal)')
plt.ylabel('Quantidade de Bebidas')
plt.show()
```

```python

plt.figure(figsize=(6, 6))
plt.boxplot(df['Açúcares totais (g)'])
plt.title('Distribuição dos Açúcares Totais (g/100ml)')
plt.ylabel('Açúcares Totais (g)')
plt.xticks([])
plt.show()
```
