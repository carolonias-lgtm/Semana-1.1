```python
import pandas as pd
dados = pd.read_excel('Dados Tarefa3.xlsx', header=1)
dados
```

```python
freq_Cerveja = dados['Preferência de bebida'].value_counts()
freq_Cerveja
```

```python
freq_Cerveja_rel = dados['Preferência de bebida'].value_counts(normalize=True) * 100
freq_Cerveja_rel
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
sns.set(style='whitegrid')
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=dados, x="Preferência de bebida", color="green")
plt.title("Preferência de bebidas")
plt.xlabel("Preferência de bebida")
plt.ylabel("Frequência")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data=dados, x="Idade", bins=[18, 19, 20, 21, 22, 23, 24, 25], kde=False, color="red")
plt.title("Preferência de bebida por idade")
plt.xlabel("Idade")
plt.ylabel("Frequência")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=dados, x="Preferência de bebida", y="Idade")
plt.title("Preferência de bebida por idade e frequência de consumo semanal")
plt.xlabel("Preferência de bebida")
plt.ylabel("Idade")
plt.show()
```

#Relatório Gráficos
Foram analisados dados de pessoas entre 18 a 25 anos sobre a frequência de consumo de álcool e prefência de 3 categorias diferentes de bebidas. Contudo, foi observado que a cerveja é a mais escolhida para consumo independente da idade, e a com maior frequência de consumo semanal.
