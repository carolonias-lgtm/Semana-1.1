#9.Como você compararia o teor de gordura entre as diferentes categorias de alimentos? Que representação seria mais informativa?
**R**: Uma das formas mais eficientes de comparar o teor de gordura entre as diferentes categorias de alimentos é por meio de um gráfico de barras, como o que está logo abaixo.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('/content/usda_data.xlsx', sheet_name='Base_didatica')
df.head(10)
```

```python
media_gordura = df.groupby('Categoria')['Gordura total (g)'].mean()
media_gordura
```

```python
media_gordura = media_gordura.sort_values(ascending=False)

media_gordura
```

```python
plt.figure(figsize=(12, 6))

plt.bar(media_gordura.index, media_gordura.values, color='pink')

plt.title('Teor médio de gordura por categoria de alimento')
plt.xlabel('Categoria de alimento')
plt.ylabel('Gordura média (g)')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```
