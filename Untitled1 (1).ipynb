```python
import matplotlib.pyplot as plt
import seaborn as sns

# Contar a frequência de cada fruta
fruit_preferences = df['Qual dessas frutas você prefere para suco natural?'].value_counts().reset_index()
fruit_preferences.columns = ['Fruta', 'Contagem']

# Criar o gráfico de barras
plt.figure(figsize=(3, 7))
sns.barplot(x='Fruta', y='Contagem', data=fruit_preferences, palette='viridis')
plt.title('Preferência de Frutas para Suco Natural')
plt.xlabel('Fruta')
plt.ylabel('Número de Respostas')
plt.xticks(rotation=90, ha='right') # Rotacionar rótulos para melhor legibilidade
plt.tight_layout()
plt.show()
```
