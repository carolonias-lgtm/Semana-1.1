# Atividade: Análise da Linha de Produção de Farinha

Considerando que o peso líquido das embalagens de farinha segue uma distribuição Normal com média $\mu = 1005\text{ g}$ e desvio-padrão $\sigma = 8\text{ g}$:

---

### Item 1: Distribuição de probabilidade e parâmetros

A variável aleatória $X$ (peso líquido em gramas) é modelada por:

$$X \sim N(\mu = 1005, \sigma^2 = 64)$$

- **Média ($\mu$):** $1005\text{ g}$
- **Desvio-padrão ($\sigma$):** $8\text{ g}$
- **Variância ($\sigma^2$):** $64\text{ g}^2$

```python
# Importação das bibliotecas necessárias
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# --- ITEM 1 ---
# Definição dos parâmetros da distribuição Normal
mu = 1005  # Média em gramas
sigma = 8  # Desvio-padrão em gramas

print("--- ITEM 1 ---")
print(f"Distribuição de X: N(mu = {mu}, sigma^2 = {sigma**2})")
print(f"Parâmetros: Média (mu) = {mu} g | Desvio-padrão (sigma) = {sigma} g")
```

### Item 2: Probabilidade $P(X \le 1000)$

Calculando o valor $Z$ para a Normal Padrão:

$$Z = \frac{1000 - 1005}{8} = -0,625$$

$$P(X \le 1000) = P(Z \le -0,625) \approx 0,2660 \text{ (26,60\%)}$$

**Resposta:** A probabilidade de uma embalagem apresentar peso inferior ou igual a 1.000 g é de **26,60%**.

```python
# --- ITEM 2 ---
# P(X <= 1000)
x_item2 = 1000
p_item2 = stats.norm.cdf(x_item2, loc=mu, scale=sigma)

print("--- ITEM 2 ---")
print(f"P(X <= {x_item2}) = {p_item2:.4f} ou {p_item2 * 100:.2f}%")
```

### Item 3: Representação gráfica de $P(X \le 1000)$

Abaixo geramos o gráfico da curva densidade de probabilidade $N(1005, 8^2)$ destacando a área equivalente a $P(X \le 1000)$.

```python
# --- ITEM 3 ---
# Gráfico de P(X <= 1000)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 4.5))
plt.plot(x, y, label='N(1005, 8²)', color='blue')

# Destacar a região X <= 1000
x_fill = np.linspace(mu - 4*sigma, 1000, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color='red', alpha=0.4, label='P(X ≤ 1000) ≈ 26,6%')

plt.axvline(1000, color='red', linestyle='--', label='X = 1000 g')
plt.title('Item 3: Distribuição de X com destaque para P(X ≤ 1000)')
plt.xlabel('Peso (g)')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Item 4: Probabilidade $P(X \ge 1015)$

Calculando o valor $Z$ para a Normal Padrão:

$$Z = \frac{1015 - 1005}{8} = 1,25$$

$$P(X \ge 1015) = P(Z \ge 1,25) = 1 - P(Z < 1,25) = 1 - 0,8944 = 0,1056 \text{ (10,56\%)}$$

**Resposta:** A probabilidade de uma embalagem apresentar peso superior ou igual a 1.015 g é de **10,56%** (notação: $P(X \ge 1015)$).

```python
# --- ITEM 4 ---
# P(X >= 1015)
x_item4 = 1015
# P(X >= 1015) = 1 - P(X < 1015) ou usando a sf (survival function)
p_item4 = stats.norm.sf(x_item4, loc=mu, scale=sigma)

print("--- ITEM 4 ---")
print(f"P(X >= {x_item4}) = {p_item4:.4f} ou {p_item4 * 100:.2f}%")
```

### Item 5: Representação gráfica de $P(X \ge 1015)$

Abaixo geramos o gráfico da curva densidade de probabilidade destacando a área equivalente a $P(X \ge 1015)$.

```python
# --- ITEM 5 ---
# Gráfico de P(X >= 1015)
plt.figure(figsize=(8, 4.5))
plt.plot(x, y, label='N(1005, 8²)', color='blue')

# Destacar a região X >= 1015
x_fill4 = np.linspace(1015, mu + 4*sigma, 500)
y_fill4 = stats.norm.pdf(x_fill4, loc=mu, scale=sigma)
plt.fill_between(x_fill4, y_fill4, color='orange', alpha=0.5, label='P(X ≥ 1015) ≈ 10,56%')

plt.axvline(1015, color='orange', linestyle='--', label='X = 1015 g')
plt.title('Item 5: Distribuição de X com destaque para P(X ≥ 1015)')
plt.xlabel('Peso (g)')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Item 6: Determinação do peso $x$ para $P(X \le x) = 0,90$

Buscamos o percentil 90 da distribuição:

$$P(Z \le z) = 0,90 \implies z \approx 1,28155$$

Convertendo para a escala de $X$:

$$x = \mu + z \cdot \sigma = 1005 + (1,28155 \cdot 8) \approx 1015,25\text{ g}$$

**Resposta:** O valor de peso $x$ é aproximadamente **1015,25 g**.

```python
# --- ITEM 6 ---
# Valor de x tal que P(X <= x) = 0.90 (Percentil 90)
p_desejada = 0.90
x_item6 = stats.norm.ppf(p_desejada, loc=mu, scale=sigma)

print("--- ITEM 6 ---")
print(f"O valor de x tal que P(X <= x) = 0.90 é: {x_item6:.2f} g")
```

### Item 7: Interpretação do resultado no contexto da linha de produção

**Interpretação:**
Exatamente **90% das embalagens** produzidas apresentam peso menor ou igual a **1015,25 g** (e apenas 10% excedem esse valor).

Na prática industrial, essa métrica ajuda no controle de custos e desperdício de matéria-prima (sobrepeso/doação de produto), permitindo recalibrar o processo se necessário.

```python
# --- ITEM 7 ---
print("--- ITEM 7 ---")
print(f"Interpretação: Exatamente 90% das embalagens produzidas nesta linha")
print(f"apresentam peso menor ou igual a {x_item6:.2f} g. Isso indica que apenas 10%")
print(f"das embalagens excedem esse valor, sendo útil para controle de custos e desperdício.")
```
